```python
numeric_cols = relevant_df.select_dtypes(include=['number']).columns
categorical_cols = relevant_df.select_dtypes(include=['object', 'category']).columns
```

```python
cat_mode = relevant_df[categorical_cols].mode()

if not cat_mode.empty:
    relevant_df[categorical_cols] = relevant_df[categorical_cols].fillna(cat_mode.iloc[0])
else:
    relevant_df[categorical_cols] = relevant_df[categorical_cols].fillna("Missing")
```

```python
relevant_df.isna().sum().sort_values(ascending=False)
```

```python
Q1 = relevant_df[numeric_cols].quantile(0.25)
Q3 = relevant_df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1
```

```python
outlier_mask = (relevant_df[numeric_cols] < (Q1 - 1.5 * IQR)) | \
               (relevant_df[numeric_cols] > (Q3 + 1.5 * IQR))

outlier_mask.sum().sort_values(ascending=False)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

relevant_df.loc[:, numeric_cols] = relevant_df[numeric_cols].clip(lower_bound, upper_bound, axis=1)
```
