```python
occupation_counts = df['Q6'].value_counts()

display(occupation_counts.head(20))

plt.figure(figsize=(10, 8))

occupation_counts.head(20).sort_values().plot(kind='barh')

plt.xlabel("Number of Respondents")
plt.ylabel("Occupation")
plt.title("Top 20 Occupations in AJQ")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

df['Q4'].value_counts().sort_index().plot(kind='bar')

plt.xlabel("Job Satisfaction (Q4)")
plt.ylabel("Number of Respondents")
plt.title("Distribution of Job Satisfaction")
plt.xticks(rotation=0)

plt.show()

q4_percent = (
    df['Q4']
    .value_counts(normalize=True)
    .sort_index() * 100
)

display(q4_percent)
```

```python
job_quality_vars = [
    'Q18', 'Q19',
    'Q20', 'Q21', 'Q22', 'Q23', 'Q24',
    'Q25A', 'Q25B', 'Q25C', 'Q25D',
    'Q31A', 'Q31B', 'Q31C',
    'Q32A', 'Q32B', 'Q32C',
    'Q33A', 'Q33B', 'Q33C', 'Q33D', 'Q33E',
    'Q33F', 'Q33G', 'Q33H', 'Q33I', 'Q33J',
    'Q34',
    'Q35A', 'Q35B', 'Q35C', 'Q35D',
    'Q36',
    'Q40',
    'Q41'
]

job_quality_vars = [col for col in job_quality_vars if col in X.columns]

display(X[job_quality_vars].head())
```

```python
def plot_category_counts(df, column, figsize=(8, 5)):
    counts = df[column].value_counts(dropna=False)

    plt.figure(figsize=figsize)
    counts.plot(kind='bar')

    plt.title(f"Distribution of Q33A: You have the freedom to decide how you do your work.")
    plt.xlabel(column)
    plt.ylabel("Number of Respondents")
    plt.xticks(rotation=45, ha='right')

    plt.tight_layout()
    plt.show()
    
plot_category_counts(df, 'Q33A')
```

```python
ct = pd.crosstab(
    df['Q33A'],
    df['Q4'],
    normalize='index'
) * 100

display(ct)

ct.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.title("Job Satisfaction by Autonomy")
plt.xlabel("Autonomy (Q33A)")
plt.ylabel("Percentage of Respondents")
plt.legend(title="Q4 Job Satisfaction")

plt.tight_layout()
plt.show()
```
