```python
import pandas as pd
wv = pd.read_excel("../../data/Work Values.xlsx")

print(wv.shape)

print(wv.head())

print(wv["Scale ID"].value_counts())
```

```python
wv = wv[wv["Scale ID"] == "EX"].copy()
print(wv.shape)  # should be (5244, 9)
```

```python
wv_wide = wv.pivot(index=["O*NET-SOC Code", "Title"], columns="Element Name", values="Data Value").reset_index()
wv_wide.columns.name = None
```

```python
print(wv_wide.shape)  
wv_wide.head()
```

```python
print(wv_wide.isnull().sum())
```

```python
wv_wide.to_csv("../../notebooks/formatting_onet/formatted_work_values.csv", index=False)
```
